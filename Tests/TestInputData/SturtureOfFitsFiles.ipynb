## 1. Chandra ACIS

### 1.1 Structure of the pi fits file

```python
fits_file = "Dash2_acisf14571_grpSNR3.pi"

# Print the extension summary
fits.info(fits_file)

with fits.open(fits_file) as hdul:
    for index, hdu in enumerate(hdul):
        print("=" * 80)
        ext_name = hdu.name if hdu.name else "No Name"
        print(f" EXTENSION {index} ({ext_name}) HEADER ".center(80, "="))
        print("=" * 80)
        
        # This breaks the header down card-by-card and forces a newline (\n)
        for card in hdu.header.cards:
            print(card)
            
        print("\n")  # Add spacing between extensions
```

### 1.2 Creating a pi file with inexisting arf, rmf, etc. 

```python
original_file = "Dash2_acisf14571_grpSNR3.pi"
new_file = "ChandraACIS_withWrongAnciliaryFileKeywords.pi"

# Open the original file in read-only mode to prevent accidental changes
with fits.open(original_file) as hdul:
    
    # 1. Modify keywords in the desired extensions
    # (Example: Modifying keywords in the Primary header [0])
    hdul[1].header['BACKFILE'] = 'InexistingBackFile'
    hdul[1].header['RESPFILE'] = 'InexistingRmfFile'
    hdul[1].header['ANCRFILE'] = 'InexistingArfFile'
    
    # 2. Write the entire modified HDU list out to the new file
    hdul.writeto(new_file, overwrite=True)
```

```python
with fits.open(new_file) as hdul:
    for index, hdu in enumerate(hdul):
        print("=" * 80)
        ext_name = hdu.name if hdu.name else "No Name"
        print(f" EXTENSION {index} ({ext_name}) HEADER ".center(80, "="))
        print("=" * 80)
        
        # This breaks the header down card-by-card and forces a newline (\n)
        for card in hdu.header.cards:
            print(card)
            
        print("\n")  # Add spacing between extensions
```
